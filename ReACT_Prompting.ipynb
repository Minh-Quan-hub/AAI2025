Code Generation with ReACT Prompting Using ChatGPT

**Prompt:**

You are a Python coding agent, and your goal is to create a Python program that allows users to manage a list of tasks. The program should allow users to add, view, complete, and delete tasks. Each task should have a unique ID, a description, and a status of either Pending or Completed.

Please make sure that the Python code can run in Google Colab. The program should accept actual user input through a simple menu. Handle empty task names, invalid menu choices, invalid task IDs, and attempts to complete or delete tasks that do not exist. Make sure to display the task list in a clear and readable format.

Use the following ReACT process:
- Plan: Briefly list the programming steps, validation rules, and edge cases.
- Generate: Write the complete Python code.
- Run: Execute the code in Google Colab and test it using actual user input.
- Observe: Review the output or any error messages. Do not assume that the code worked.
- Fix: If an error or weakness appears, revise the code and explain the change briefly. Run the revised code again.

```python
"""A simple interactive task manager for Python and Google Colab."""

TABLE_WIDTH = 50

def display_tasks(tasks):
    """Display all tasks in a readable table."""
    if not tasks:
        print("\nNo tasks have been added yet.")
        return

    print("\nCurrent Tasks")
    print("-" * TABLE_WIDTH)
    print(f"{'ID':<6}{'Description':<35}{'Status'}")
    print("-" * TABLE_WIDTH)

    for task in tasks:
        print(f"{task['id']:<6}{task['description']:<35}{task['status']}")

    print("-" * TABLE_WIDTH)


def add_task(tasks, next_id):
    """Add a task after checking that its name is not empty."""
    description = input("Enter the task name: ").strip()

    if not description:
        print("Task name cannot be empty.")
        return next_id

    tasks.append({
        "id": next_id,
        "description": description,
        "status": "Pending",
    })

    print(f"Task {next_id} added successfully.")
    return next_id + 1


def get_task_id():
    """Read and validate a positive integer task ID."""
    task_id_text = input("Enter the task ID: ").strip()

    try:
        task_id = int(task_id_text)
    except ValueError:
        print("Invalid task ID. Please enter a whole number.")
        return None

    if task_id <= 0:
        print("Invalid task ID. Please enter a positive whole number.")
        return None

    return task_id


def complete_task(tasks):
    """Mark an existing task as completed."""
    task_id = get_task_id()

    if task_id is None:
        return

    for task in tasks:
        if task["id"] == task_id:
            if task["status"] == "Completed":
                print(f"Task {task_id} is already completed.")
            else:
                task["status"] = "Completed"
                print(f"Task {task_id} marked as completed.")
            return

    print(f"No task exists with ID {task_id}.")


def delete_task(tasks):
    """Delete an existing task."""
    task_id = get_task_id()

    if task_id is None:
        return

    for index, task in enumerate(tasks):
        if task["id"] == task_id:
            deleted_description = tasks.pop(index)["description"]
            print(f"Task {task_id} ({deleted_description}) deleted.")
            return

    print(f"No task exists with ID {task_id}.")


def display_menu():
    """Display the available actions."""
    print("\nTask Manager")
    print("1. Add a task")
    print("2. View tasks")
    print("3. Complete a task")
    print("4. Delete a task")
    print("5. Exit")


def main():
    """Run the task manager until the user chooses Exit."""
    tasks = []
    next_id = 1

    print("Welcome to the Task Manager!")

    while True:
        display_menu()
        choice = input("Choose an option (1-5): ").strip()

        if choice == "1":
            next_id = add_task(tasks, next_id)
        elif choice == "2":
            display_tasks(tasks)
        elif choice == "3":
            complete_task(tasks)
        elif choice == "4":
            delete_task(tasks)
        elif choice == "5":
            print("Goodbye!")
            break
        else:
            print("Invalid menu choice. Please select a number from 1 to 5.")


if __name__ == "__main__":
    main()
```
