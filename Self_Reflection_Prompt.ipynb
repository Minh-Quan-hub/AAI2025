Code generation with Self-Reflection Prompt using ChatGPT

**Prompt:**

Build a study assistant bot using gemini API key and the goal is to create a clear study summary from the note that student provided.

You first will need to write a draft summary and the review the summary using these requirements:
- Explain the main topic accurately.
- Include the definition, the key characteristics, and one real-world example.
- Use simple language for a beginner.
- Keep the summary clear and within 100 and 120 words.
- Format it as three bullet points followed by one takeaway sentence.
- Do not add information that is not included in the source passage.

After the review, write an improved summary that satisfies all the requirements.

Clearly label your response:
1. Draft Summary
2. Self-Critique
3. Improved Summary

```python
from google import genai
from getpass import getpass
import google.generativeai as genai
from google.colab import userdata

# Load the Gemini API key from Google Colab Secrets
try:
    api_key = userdata.get("GEMINI_API_KEY")
    genai.configure(api_key=api_key)
except Exception:
    raise ValueError(
        "Gemini API key not found. Add GEMINI_API_KEY to Colab Secrets."
    )

MODEL_NAME = "gemini-3.6-flash"


def study_assistant(student_notes):
    """
    Creates a study summary from notes provided by the student.
    The assistant will:
    1. Write a draft summary
    2. Review the draft
    3. Write an improved summary
    """

    prompt = f"""
You are a Study Assistant. Your goal is to create a clear study summary
from notes provided by a student.

SOURCE PASSAGE:
{student_notes}

Follow this process:

STEP 1 - Draft Summary
Write a first draft of the summary.

The summary should:
- Explain the main topic accurately.
- Include the definition, key characteristics, and one real-world example.
- Use simple language for a beginner.
- Is clear and within 100 and 120 words.
- Format the summary as exactly three bullet points followed by one takeaway sentence.
- Do not add information that is not included in the source passage.

STEP 2 - Self-Critique
Review your draft summary.

Check whether the draft:
- Accurately explains the main topic.
- Includes the definition.
- Includes the key characteristics.
- Includes one real-world example.
- Uses simple language.
- Is clear and within 100 and 120 words.
- Contains exactly three bullet points and one takeaway sentence.
- Uses only information from the source passage.

Briefly explain in a short paragraph what was done well and what needs improvement.

STEP 3 - Improved Summary
Using your self-critique, rewrite the summary so that it satisfies all
of the requirements.

Clearly label your response exactly as:

1. Draft Summary
2. Self-Critique
3. Improved Summary

Use "•" for all bullet points.
Do not use Markdown asterisks (* or **).
Do not use "***" as section dividers.
"""

    # Send prompt to Gemini
    response = genai.GenerativeModel(
        "gemini-3.6-flash",
    ).generate_content(contents=prompt)

    return response.text


# -----------------------------
# Get notes from the student
# -----------------------------

print("=== Study Assistant ===")
print("Paste your study notes below.")
print("Type END on a new line when you are finished.\n")

lines = []

while True:
    line = input()

    if line.strip().upper() == "END":
        break

    lines.append(line)

student_notes = "\n".join(lines).strip()


# Check for empty notes
if not student_notes:
    print("\nError: Please provide study notes before generating a summary.")

else:
    print("\nGenerating study summary...\n")

    try:
        result = study_assistant(student_notes)

        print("=" * 60)
        print("STUDY ASSISTANT RESULT")
        print("=" * 60)
        print(result)

    except Exception as error:
        print("An error occurred:", error)
```
